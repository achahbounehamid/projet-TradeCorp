```python
# Q1 — Créer la SparkSession
```

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder \
    .appName("TradeCorp ETL") \
    .getOrCreate()
print("Version de Spark :", spark.version)
```

```python
# Q2 — Lire les 8 CSV
```

```python
# 1. le chemin vers le dossier contenant les fichiers
PATH = "/home/jovyan/data/"
# 2. On charge les 8 fichiers
df_customers = spark.read.csv(PATH + "customers.csv", header=True, inferSchema=True)
df_categories = spark.read.csv(PATH + "categories.csv", header=True, inferSchema=True)
df_employees = spark.read.csv(PATH + "employees.csv", header=True, inferSchema=True)
df_order_details = spark.read.csv(PATH + "order_details.csv", header=True, inferSchema=True)
df_orders = spark.read.csv(PATH + "orders.csv", header=True, inferSchema=True)
df_products = spark.read.csv(PATH + "products.csv", header=True, inferSchema=True)
df_shippers = spark.read.csv(PATH + "shippers.csv", header=True, inferSchema=True)
df_suppliers = spark.read.csv(PATH + "suppliers.csv", header=True, inferSchema=True)

```

```python
#Q10 — Colonnes et types
```

```python
import pandas as pd
import time

#************PANDAS**********************

# 1. Enregistrer le temps de départ
debut_pd = time.time()

# 2. Exécuter la lecture du CSV
df_pandas = pd.read_csv(PATH + "orders.csv")

# 3. Enregistrer le temps de fin
fin_pd = time.time()

# 4. Afficher le résultat et les informations
print(f"Temps Pandas : {fin_pd - debut_pd:.4f} secondes")
print(df_pandas.head())
print(df_pandas.info())

#**********************Sparks*************
# 1. Enregistrer le temps de départ
debut_sp = time.time()
# 2. Exécuter la lecture du CSV
df_orders = spark.read.csv(PATH + "orders.csv", header=True, inferSchema=True)
# 3. Forcer le calcul réel de Spark (Action)
df_orders.count()
# 4. Enregistrer le temps de fin
fin_sp = time.time()
# 4. Afficher le résultat et les informations
print(f"Temps Spark  : {fin_sp - debut_sp:.4f} secondes")
df_orders.show(5)
df_orders.printSchema()

Explication des résultats (Q9)
#1. Comparaison des temps d'exécution : plus rapide que Spark sur ce fichier.

#Pandas : 0.0082 seconde

#Spark : 0.2298 seconde 
#orders.csv est très petit (830 lignes). Pandas traite les données directement en mémoire locale.Spark initialisation du plan d'exécution DAG et gestion du cluster qui  prend plus de temps que le calcul lui-même.

```
