```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder \
    .appName("TradeCorp ETL") \
    .getOrCreate()

PATH="/home/jovyan/data/tmp/"
df_customers = spark.read.parquet(PATH + "customers")
df_categories= spark.read.parquet(PATH + "categories")
df_employees= spark.read.parquet(PATH + "employees")
df_order_details= spark.read.parquet(PATH + "order_detail")
df_orders = spark.read.parquet(PATH + "orders")
df_products = spark.read.parquet(PATH + "products")
df_shippers = spark.read.parquet(PATH + "shippers")
df_suppliers = spark.read.parquet(PATH + "suppliers")

dfs = {
    "customers": df_customers,
    "categories": df_categories,
    "employees": df_employees,
    "order_details": df_order_details,
    "orders": df_orders,
    "products": df_products,
    "shippers": df_shippers,
    "suppliers": df_suppliers
}
for nom_df, df in dfs.items():
    print(f"\n=== {nom_df.upper()} ===")
    for c in df.columns:
        nb_nulls = df.filter(col(c).isNull()).count()
        print(f"Colonne {c} : {nb_nulls} valeur(s) nulle(s)")
```

```python
from pyspark.sql.functions import col, trim, initcap, upper

# Définition directe dans la cellule
def clean_customers(df):
    clean_customer = df \
        .withColumn("company_name", trim(col("company_name"))) \
        .withColumn("contact_name", initcap(col("contact_name"))) \
        .withColumn("country", upper(col("country"))) \
        .dropDuplicates(["customer_id"])
    return clean_customer

# Exécution
df_customers_clean = clean_customers(df_customers)
df_customers_clean.show(5)
```

```python
from pyspark.sql.functions import col, round

def clean_order_details(df):
    # Gestion souple selon le nom initial des colonnes
    if "prix_unitaire" in df.columns:
        df_order_detail = df \
            .withColumn("prix_unitaire", col("prix_unitaire").cast("double")) \
            .withColumn("quantite", col("quantite").cast("integer")) \
            .withColumn("discount", col("discount").cast("double"))
    else:
        df_order_detail = df \
            .withColumn("unit_price", col("unit_price").cast("double")) \
            .withColumn("quantity", col("quantity").cast("integer")) \
            .withColumn("discount", col("discount").cast("double")) \
            .withColumnRenamed("unit_price", "prix_unitaire") \
            .withColumnRenamed("quantity", "quantite")
    return df_order_detail

def add_sous_total(df):
    return df.withColumn("sous_total", round(col("prix_unitaire") * col("quantite") * (1 - col("discount")), 2))

df_details_clean = clean_order_details(df_order_details)
df_details_clean = add_sous_total(df_details_clean)
df_details_clean.show(5)
```

```python
from pyspark.sql.functions import col, concat, lit

def clean_employees(df):
    df_clean_employee = df \
        .select("employee_id", "first_name", "last_name", "title", "hire_date", "city", "country") \
        .withColumn("full_name", concat(col("first_name"), lit(" "), col("last_name")))
    return df_clean_employee

df_employees_clean = clean_employees(df_employees)
df_employees_clean.show(5)
```

```python
from pyspark.sql.functions import col

def clean_products(df):
    df_clean_product = df \
        .withColumn("unit_price", col("unit_price").cast("double")) \
        .withColumn("en_stock", col("units_in_stock") > 0)
    return df_clean_product

df_products_clean = clean_products(df_products)
df_products_clean.show(5)
```
