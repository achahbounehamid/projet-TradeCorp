```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder \
    .appName("TradeCorp ETL") \
    .getOrCreate()

PATH="/home/jovyan/data/tmp/"
df_customers = spark.read.parquet(PATH + "customers")
df_categories= spark.read.parquet(PATH + "categories")
df_employees= spark.read.parquet(PATH + "employees")
df_order_details= spark.read.parquet(PATH + "order_detail")
df_orders = spark.read.parquet(PATH + "orders")
df_products = spark.read.parquet(PATH + "products")
df_shippers = spark.read.parquet(PATH + "shippers")
df_suppliers = spark.read.parquet(PATH + "suppliers")

dfs = {
    "customers": df_customers,
    "categories": df_categories,
    "employees": df_employees,
    "order_details": df_order_details,
    "orders": df_orders,
    "products": df_products,
    "shippers": df_shippers,
    "suppliers": df_suppliers
}
for nom_df, df in dfs.items():
    print(f"\n=== {nom_df.upper()} ===")
    for c in df.columns:
        nb_nulls = df.filter(col(c).isNull()).count()
        print(f"Colonne {c} : {nb_nulls} valeur(s) nulle(s)")
```
